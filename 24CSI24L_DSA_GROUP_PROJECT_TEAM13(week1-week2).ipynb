# 1. Import Required Libraries and Load the Dataset

**Purpose:** Import the libraries required for data analysis and visualization and Load the raw dataset into a Pandas DataFrame for analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\ELCOT\Documents\SmartFarming_recommendation.csv")
```

# 3. Check Dataset Shape and Check Column Names

**Purpose:** Identify the number of rows and columns in the dataset and Identify all available features in the dataset.

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())
```

# 4. Check Dataset Information

**Purpose:** Understand the dataset structure, data types, and non-null values.

```python
df.info()
```

# 5. Identify Feature Types

**Purpose:** Separate numerical and categorical features for further analysis.

```python
numerical_cols = df.select_dtypes(
    include=np.number
).columns.tolist()

categorical_cols = df.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

print("Numerical Features:")
print(numerical_cols)

print("\nCategorical Features:")
print(categorical_cols)
```

# 6. Check Missing Values

**Purpose:** Identify missing values and assess data completeness.

```python
df.isnull().sum()
```

# 7. Check Duplicate Records

**Purpose:** Identify duplicate observations in the dataset.

```python
df.duplicated().sum()
```

# 8. Check Unique Values

**Purpose:** Identify the number of unique values in each feature.

```python
df.nunique()
```

# 9. Statistical Summary

**Purpose:** Understand the basic statistical characteristics of numerical features.

```python
df.describe(include="all")
```

# 10. Check Constant Columns

**Purpose:** Identify columns with only one unique value.

```python
constant_columns = [
    col for col in df.columns
    if df[col].nunique() == 1
]

print(constant_columns)
```

# 11. Data Quality Summary

**Purpose:** Summarize the quality and characteristics of each feature.

```python
data_quality_report = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isnull().sum(),
    "Missing %": ((df.isnull().sum() / len(df)) * 100).round(2),
    "Unique Values": df.nunique()
})

data_quality_report
```
